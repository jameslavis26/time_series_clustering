```python
import numpy as np
import matplotlib.pyplot as plt
from time_series.data_generators import LorenzGenerator
```

```python
t, X1 = LorenzGenerator(
   rho=28, 
   sigma=10, 
   beta=8/3,
   x0=np.array([1, 1, 1]), 
   T=10,
   dt=0.01,
   noise_covariance=0
)()

t, X2 = LorenzGenerator(
   rho=28, 
   sigma=10, 
   beta=8/3,
   x0=np.array([1, -1, -1]), 
   T=10,
   dt=0.01,
   noise_covariance=0
)()
```

```python
fig, ax = plt.subplots(ncols=2, figsize=(10, 8), subplot_kw={'projection': '3d'})

ax[0].plot(*X1.T)
ax[0].set_title("Curve with x0 = [1, 1, 1]")
ax[1].plot(*X2.T)
ax[1].set_title("Curve with x0 = [1, -1, -1]")
```

```python
fig, ax = plt.subplots(ncols=1, figsize=(10, 8), subplot_kw={'projection': '3d'})

ax.plot(*X1.T)
ax.plot(*X2.T)
```
