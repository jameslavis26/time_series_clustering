```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("~/Downloads/Splitwise expenses Jan 23.csv")
```

```python
df.head(20)
```

```python
df = df[df["Cost"] != " "]

df.index = pd.to_datetime(df["Date"])
df["Cost"] = df["Cost"].astype(float)
```

```python
df["Month"] = df.index.month
df["Year"] = df.index.year

df.loc[df["Category"] == "Taxi", "Category"] = "Dining out"
df.loc[df["Category"] == "Household supplies", "Category"] = "Dining out"
df.loc[df["Description"].str.lower().str.contains("dm"), "Category"] = "Groceries"
df.loc[df["Description"] == "Vacuum ", "Category"] = "General" 

df.loc[df["Description"].isin(
    [
        "Frittenwerk",
        "Maccas",
        "Maccas ",
        "Asian food"
    ]
), "Category"] = "Dining out"

df.loc[df["Description"].isin(
    [
        "Presents Mama&Papa",
        "Wine for Neils"
    ]
), "Category"] = "Gifts"

df.loc[df["Description"].isin(["Thrift store"]), "Category"] = "Misc"

misc_categories = [
    'TV/Phone/Internet',
    'Movies', 
    'Bus/train', 
    'Sports', 
    'Liquor',
]

df.loc[df["Description"].isin(misc_categories), "Category"] = "Misc"
```
